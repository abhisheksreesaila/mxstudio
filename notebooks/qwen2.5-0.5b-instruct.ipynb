## What is this model?

Think of this checkpoint as one point on a design trade-off curve: parameter count,

## What's the innovation?

Qwen2.5 pushed small-model quality by pretraining on ~18T tokens with aggressive data filtering, then applying the same post-training recipe (supervised finetuning + multi-stage reinforcement learning) used by the flagship models. The surprising result: even a 0.5B checkpoint keeps 32K context and usable instruction following — proof that data quality and post-training transfer down the size ladder.

### Key ideas to take away

- Small models can inherit the post-training recipe of larger siblings
- Long context (32K) is no longer reserved for huge models
- Instruction tuning matters as much as raw parameter count at this size


## Architecture & training (plain language)

Dense decoder-only transformer (Qwen2 architecture): grouped-query attention (GQA), SwiGLU feed-forward layers, RoPE positional embeddings, and a large vocabulary (~150k) tokenizer that helps multilingual and code performance. The Instruct variant is chat-tuned with a ChatML-style template.

### When should you pick this model?

Use this when you want a fast local sandbox: UI demos, notebook labs, API integration tests, or teaching the OpenAI chat protocol without waiting on a large download. Do not expect strong multi-step reasoning or complex coding — graduate to 1.5B/3B for that.

### 1. Connect

Create an OpenAI client pointed at the local MAX server and list the loaded model.


````python
from openai import OpenAI

MODEL = "Qwen/Qwen2.5-0.5B-Instruct"
BASE = "http://localhost:8100/v1"
client = OpenAI(base_url=BASE, api_key="EMPTY")

models = [m.id for m in client.models.list()]
print("endpoint:", BASE)
print("available:", models)
assert any(MODEL in m or m in MODEL for m in models) or models, "Server returned no models — is Serve running?"
````

### 2. Ask the model to teach you

First completion: have the model explain itself. Small models can be wrong —
treat the answer as a *starting point*, then compare with the innovation section above.


````python
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {
            "role": "user",
            "content": (
                "You are helping a student. In 4 short bullet points, explain: "
                "(1) what kind of model you are, (2) roughly how large you are, "
                "(3) one thing you are good at, (4) one limitation they should remember. "
                "Be honest; do not invent papers."
            ),
        }
    ],
    max_tokens=220,
    temperature=0.4,
)
print(response.choices[0].message.content)
````

### 3. Streaming tokens

Streaming returns tokens as they are generated. That is what makes chat UIs feel live —
and it is how we will measure time-to-first-token later.


````python
stream = client.chat.completions.create(
    model=MODEL,
    messages=[{
        "role": "user",
        "content": "In one short paragraph, what is 'post-training' for LLMs, and why does it matter for small models?",
    }],
    max_tokens=160,
    temperature=0.5,
    stream=True,
)
for chunk in stream:
    if chunk.choices and chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end="", flush=True)
print()
````

### 4. Temperature playground

**Temperature** controls randomness. `0.0` ≈ deterministic; higher values explore more.
Same prompt, three temperatures — watch how the answers diverge.


````python
prompt = (
    "Invent a name and a one-line pitch for an AI-powered study tool "
    "that helps students read model papers."
)
for temp in (0.0, 0.7, 1.2):
    r = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=temp,
        max_tokens=80,
    )
    print(f"--- temperature={temp}")
    print(r.choices[0].message.content.strip())
    print()
````

### 5. Concept check (prompt the model, then judge it)

Ask a precise question tied to *this* model's story. Then grade the answer against
the **What's the innovation?** section — did it miss data scale, openness, or MoE routing?


````python
r = client.chat.completions.create(
    model=MODEL,
    messages=[{"role": "user", "content": "According to your training story, what is special about Qwen2.5's recipe for small models? Mention pretraining scale and post-training in plain language."}],
    max_tokens=220,
    temperature=0.3,
)
print(r.choices[0].message.content)
print("\n---\nNow compare this answer with the innovation section at the top of the notebook.")
````

### 6. Mini benchmark

Rough local throughput: **TTFT** (time to first token) and **tok/s** (decode speed).
Numbers vary with GPU load — the point is to build intuition, not publish a leaderboard.


````python
import time

t0 = time.perf_counter()
first = None
tokens = 0
stream = client.chat.completions.create(
    model=MODEL,
    messages=[{"role": "user", "content": "Count from 1 to 40, separated by commas."}],
    max_tokens=120,
    temperature=0.0,
    stream=True,
)
for chunk in stream:
    if chunk.choices and chunk.choices[0].delta.content:
        if first is None:
            first = time.perf_counter()
        tokens += 1
assert first is not None, "no tokens received"
gen_time = time.perf_counter() - first
print(f"TTFT: {first - t0:.3f}s")
print(f"Decode: {tokens} tokens in {gen_time:.2f}s -> {tokens / gen_time:.1f} tok/s")
````
