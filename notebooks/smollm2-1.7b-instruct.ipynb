## What is this model?

## What's the innovation?

SmolLM2 is a study in data-centric training: ~11T tokens with new curated datasets (FineMath, Stack-Edu, SmolTalk) and multi-stage data rebalancing. Unlike many 'open weights' releases, the paper and datasets document the full recipe — making it one of the most reproducible small models available.

### Key ideas to take away

- Open weights ≠ open science — SmolLM2 aims for both
- Dataset curation (math, education, chat) beats brute-force scraping alone
- Side-by-side eval with Qwen teaches more than either model alone


## Architecture & training (plain language)

Dense decoder-only transformer in the SmolLM lineage. The Instruct variant is tuned on SmolTalk and related chat/instruction mixtures. Context is 8K — shorter than Qwen2.5's 32K, which is a useful trade-off to discuss in class.

### When should you pick this model?

Teaching openness and reproducibility; comparing data recipes against Qwen; Apache 2.0 deployments where you want a fully open stack.

## Model Summary

SmolLM2 is a family of compact language models available in three size: 135M, 360M, and 1.7B parameters. They are capable of solving a wide range of tasks while being lightweight enough to run on-device. More details in our paper: https://arxiv.org/abs/2502.02737v1

The 1.7B variant demonstrates significant advances over its predecessor SmolLM1-1.7B, particularly in instruction following, knowledge, reasoning, and mathematics. It was trained on 11 trillion tokens using a diverse dataset combination: FineWeb-Edu, DCLM, The Stack, along with new mathematics and coding datasets that we curated and will release soon. We developed the instruct version through supervised fine-tuning (SFT) using a combination of public datasets and our own curated datasets.…

## Limitations

SmolLM2 models primarily understand and generate content in English. They can produce text on a variety of topics, but the generated content may not always be factually accurate, logically consistent, or free from biases present in the training data. These models should be used as assistive tools rather than definitive sources of information. Users should always verify important information and critically evaluate any generated content.

### 1. Connect

Create an OpenAI client pointed at the local MAX server and list the loaded model.


````python
from openai import OpenAI

MODEL = "HuggingFaceTB/SmolLM2-1.7B-Instruct"
BASE = "http://localhost:8100/v1"
client = OpenAI(base_url=BASE, api_key="EMPTY")

models = [m.id for m in client.models.list()]
print("endpoint:", BASE)
print("available:", models)
assert any(MODEL in m or m in MODEL for m in models) or models, "Server returned no models — is Serve running?"
````

### 2. Ask the model to teach you

First completion: have the model explain itself. Small models can be wrong —
treat the answer as a *starting point*, then compare with the innovation section above.


````python
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {
            "role": "user",
            "content": (
                "You are helping a student. In 4 short bullet points, explain: "
                "(1) what kind of model you are, (2) roughly how large you are, "
                "(3) one thing you are good at, (4) one limitation they should remember. "
                "Be honest; do not invent papers."
            ),
        }
    ],
    max_tokens=220,
    temperature=0.4,
)
print(response.choices[0].message.content)
````

### 3. Streaming tokens

Streaming returns tokens as they are generated. That is what makes chat UIs feel live —
and it is how we will measure time-to-first-token later.


````python
stream = client.chat.completions.create(
    model=MODEL,
    messages=[{
        "role": "user",
        "content": "In one short paragraph, what is 'post-training' for LLMs, and why does it matter for small models?",
    }],
    max_tokens=160,
    temperature=0.5,
    stream=True,
)
for chunk in stream:
    if chunk.choices and chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end="", flush=True)
print()
````

### 4. Temperature playground

**Temperature** controls randomness. `0.0` ≈ deterministic; higher values explore more.
Same prompt, three temperatures — watch how the answers diverge.


````python
prompt = (
    "Invent a name and a one-line pitch for an AI-powered study tool "
    "that helps students read model papers."
)
for temp in (0.0, 0.7, 1.2):
    r = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=temp,
        max_tokens=80,
    )
    print(f"--- temperature={temp}")
    print(r.choices[0].message.content.strip())
    print()
````

### 5. Concept check (prompt the model, then judge it)

Ask a precise question tied to *this* model's story. Then grade the answer against
the **What's the innovation?** section — did it miss data scale, openness, or MoE routing?


````python
r = client.chat.completions.create(
    model=MODEL,
    messages=[{"role": "user", "content": "What does it mean for a language model release to be 'fully open'? List weights, data, code, and documentation — and why each matters."}],
    max_tokens=220,
    temperature=0.3,
)
print(r.choices[0].message.content)
print("\n---\nNow compare this answer with the innovation section at the top of the notebook.")
````

### 6. Mini benchmark

Rough local throughput: **TTFT** (time to first token) and **tok/s** (decode speed).
Numbers vary with GPU load — the point is to build intuition, not publish a leaderboard.


````python
import time

t0 = time.perf_counter()
first = None
tokens = 0
stream = client.chat.completions.create(
    model=MODEL,
    messages=[{"role": "user", "content": "Count from 1 to 40, separated by commas."}],
    max_tokens=120,
    temperature=0.0,
    stream=True,
)
for chunk in stream:
    if chunk.choices and chunk.choices[0].delta.content:
        if first is None:
            first = time.perf_counter()
        tokens += 1
assert first is not None, "no tokens received"
gen_time = time.perf_counter() - first
print(f"TTFT: {first - t0:.3f}s")
print(f"Decode: {tokens} tokens in {gen_time:.2f}s -> {tokens / gen_time:.1f} tok/s")
````
